```python
import pytesseract, re
from pdf2image import convert_from_path

PDF_PATH    = "/content/drive/MyDrive/gemma 4/pdfs/Unit 10-Applied Biology.pdf"
UNIT_NUMBER = "10"
# ════════════════════════════════════

def clean_text(text):
    text = text.replace('\u200c', '')
    text = re.sub(r'€|©', '', text)
    text = re.sub(r'ජාතික අධ්‍යාපන ආයතනය.*?ඇවිරිණි\.', '', text)
    text = re.sub(r'සමිපත්\s*පොත.*?\n', '', text)
    text = re.sub(r'සම්පත්\s*පොත.*?\n', '', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    return text.strip()

print(f"⏳ Unit {UNIT_NUMBER} PDF loading...")
all_pages = convert_from_path(PDF_PATH)
print(f" {len(all_pages)} pages found .")

syllabus_data = []
print("🔍starting OCR ...")

for i, page_img in enumerate(all_pages):
    text    = pytesseract.image_to_string(page_img, lang='sin+eng')
    cleaned = clean_text(text)

    if len(cleaned) < 50:
        print(f"📄 පිටුව {i+1} — skip")
        continue

    syllabus_data.append({
        "id"     : f"u{UNIT_NUMBER}_p{i+1}",
        "unit"   : UNIT_NUMBER,
        "page"   : i + 1,
        "content": cleaned
    })
    print(f"📄 page {i+1} finished ✓")

print(f"\n✅  {len(syllabus_data)} Chunks ready")
```

```python
import os, pickle, numpy as np
import faiss
from sentence_transformers import SentenceTransformer

# ══ Config ══
DB_PATH = "/content/drive/MyDrive/gemma 4/biology_faiss_db"
os.makedirs(DB_PATH, exist_ok=True)

embed_model = SentenceTransformer('intfloat/multilingual-e5-small')
print("✅ Model ready!")

existing_index = None
existing_docs = []
existing_metas = []

if os.path.exists(f"{DB_PATH}/index.faiss"):
    print("⏳ Existing DB found. Loading it....")
    existing_index = faiss.read_index(f"{DB_PATH}/index.faiss")
    with open(f"{DB_PATH}/documents.pkl", "rb") as f:
        old_data = pickle.load(f)
        existing_docs = old_data["documents"]
        existing_metas = old_data["metadatas"]
    print(f" {len(existing_docs)} old chunks were loaded.")

print(f"⏳ Embedding {len(syllabus_data)} new chunks......")
new_documents = [item["content"] for item in syllabus_data]
new_metadatas = [{"unit": item["unit"], "page": item["page"], "id": item["id"]}
                 for item in syllabus_data]

new_embs = embed_model.encode(
    [f"passage: {d}" for d in new_documents],
    show_progress_bar=True
)

if existing_index is not None:
    existing_index.add(new_embs.astype(np.float32))
    index = existing_index
else:
    dim = new_embs.shape[1]
    index = faiss.IndexFlatL2(dim)
    index.add(new_embs.astype(np.float32))

final_docs = existing_docs + new_documents
final_metas = existing_metas + new_metadatas

faiss.write_index(index, f"{DB_PATH}/index.faiss")
with open(f"{DB_PATH}/documents.pkl", "wb") as f:
    pickle.dump({"documents": final_docs, "metadatas": final_metas}, f)

print(f"✅ DB was successfully updated!")
print(f"📊 Total Chunks: {index.ntotal}")
```

```python
import faiss, pickle, numpy as np

def load_db():
    DB_PATH = "/content/drive/MyDrive/gemma 4/biology_faiss_db"
    index   = faiss.read_index(f"{DB_PATH}/index.faiss")
    with open(f"{DB_PATH}/documents.pkl", "rb") as f:
        data = pickle.load(f)
    return index, data["documents"], data["metadatas"]

index, documents, metadatas = load_db()
print(f"✅ DB loaded! Chunks: {index.ntotal}")

def rag_query(question, top_k=3):
    q_emb = embed_model.encode(f"query: {question}")
    q_emb = np.array([q_emb]).astype(np.float32)

    distances, indices = index.search(q_emb, top_k)

    print(f"📄 Found chunks:\n")
    print("─" * 40)
    chunks = []
    for rank, idx in enumerate(indices[0]):
        meta  = metadatas[idx]
        chunk = documents[idx]
        chunks.append(chunk)
        print(f"🔹 Chunk {rank+1} — Unit {meta['unit']}, Page {meta['page']}:")
        print(chunk[:400])
        print("─" * 40)

    context = "\n\n---\n\n".join(chunks)
    prompt  = f"""පහත context එකෙන් විතරක් පිළිතුර දෙන්න.\n\nContext:\n{context}\n\nප්‍රශ්නය: {question}\n\nපිළිතුර:"""
    return prompt

# Test
prompt = rag_query("අසම්පූර්ණ ප්‍රමුඛතාවය යනු කුමක්ද? මෙන්ඩලීය සම්පූර්ණ ප්‍රමුඛතාවයෙන් එය වෙනස් වන්නේ කෙසේද?")
print("\n✅ Prompt ready!")
```
